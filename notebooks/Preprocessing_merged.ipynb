```
def check_pct_outlier(data) :
    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers   = data[(data < lower_bound) | (data > upper_bound)]
    n_outliers = data[(data >= lower_bound) & (data <= upper_bound)]
    l_outliers = data[(data < lower_bound)]
    u_outliers = data[(data > upper_bound)]
    
    outlier_percentage   = (len(outliers) / len(data)) * 100
    n_outlier_percentage = (len(n_outliers) / len(data)) * 100
    l_outlier_percentage = (len(l_outliers) / len(data)) * 100
    u_outlier_percentage = (len(u_outliers) / len(data)) * 100
    
    # print(f'Persentase Lower Outlier : %.3f' % l_outlier_percentage)
    # print(f'Persentase Upper Outlier : %.3f' % u_outlier_percentage)
    # print(f'Persentase No Outlier    : %.3f' % n_outlier_percentage)
    # print(f'Persentase Total Outlier : %.3f' % outlier_percentage)
    return outlier_percentage, n_outlier_percentage, l_outlier_percentage, u_outlier_percentage
```

```
def remove_outlier_iqr(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers    = df[col][(df[col] < lower_bound) | (df[col] > upper_bound)]
    no_outliers = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]
    
    outlier_percentage = (len(outliers) / len(df[col])) * 100
    
    print(f'Persentase Total Outlier : %.3f' % outlier_percentage)
    return no_outliers
```

**Explanation of Features in the Previous Application Dataset :**</br>
0. SK_ID_PREV </br>
ID of previous credit in Home credit related to loan in our sample. (One loan in our sample can have, or more previous loan applications in Home Credit, previous application could, but not necessarily have to lead to credit),hashed
1. SK_ID_CURR </br>
ID of loan in our sample,hashed </br>
2. NAME_CONTRACT_TYPE </br>
Contract product type (Cash loan, consumer loan [POS] ,...) of the previous application </br>
3. AMT_ANNUITY </br>
Annuity of previous application </br>
4. AMT_APPLICATION </br>
For how much credit did client ask on the previous application </br>
5. AMT_CREDIT </br>
Final credit amount on the previous application. This differs from AMT_APPLICATION in a way that the AMT_APPLICATION is the amount for which the client initially applied for, but during our approval process he could have received different amount - AMT_CREDIT </br>
6. AMT_DOWN_PAYMENT </br>
Down payment on the previous application </br>
7. AMT_GOODS_PRICE </br>
Goods price of good that client asked for (if applicable) on the previous application </br>

Approximately at what day hour did the client apply for the previous application,rounded </br>

#### D. Feature Transformation

Sebelum transformasi data numerik, dilihat terlebih dahulu distribusi data untuk menentukan metode yang tepat.

```
#Numerical features from latest data
num_col2 = prev_app2[['AMT_ANNUITY', 'AMT_APPLICATION', 'AMT_CREDIT', 'AMT_DOWN_PAYMENT', 'AMT_GOODS_PRICE',
           'DAYS_DECISION', 'DAYS_FIRST_DRAWING', 'DAYS_FIRST_DUE', 'DAYS_LAST_DUE_1ST_VERSION', 'DAYS_LAST_DUE', 'DAYS_TERMINATION']]

# Determine the distribution type of each numeric column
distribution_types = {}

for col in num_col2.columns:
    skewness = num_col2[col].skew()
    if skewness > 0.5:
        distribution_types[col] = 'Right-skewed'
    elif skewness < -0.5:
        distribution_types[col] = 'Left-skewed'
    else:
        distribution_types[col] = 'Normal'

# Display the results
for col, dist_type in distribution_types.items():
    print(f'{col}: {dist_type}')
```

### HANDLING OUTLIERS

```
print(f'Jumlah baris sebelum memfilter outlier: {len(merged_df)}')

filtered_entries = np.array([True] * len(merged_df))
for col in ['MONTHS_BALANCE', 'AMT_BALANCE', 'AMT_CREDIT_LIMIT_ACTUAL', 'AMT_DRAWINGS_ATM_CURRENT', 'AMT_DRAWINGS_CURRENT', 'AMT_DRAWINGS_OTHER_CURRENT', 'AMT_DRAWINGS_POS_CURRENT', 'AMT_INST_MIN_REGULARITY', 'AMT_PAYMENT_CURRENT', 'AMT_PAYMENT_TOTAL_CURRENT', 'AMT_RECEIVABLE_PRINCIPAL', 'AMT_RECIVABLE', 'AMT_TOTAL_RECEIVABLE', 'CNT_DRAWINGS_ATM_CURRENT', 'CNT_DRAWINGS_CURRENT', 'CNT_DRAWINGS_OTHER_CURRENT', 'CNT_DRAWINGS_POS_CURRENT', 'CNT_INSTALMENT_MATURE_CUM', 'SK_DPD', 'SK_DPD_DEF']:
    Q1 = merged_df[col].quantile(0.25)
    Q3 = merged_df[col].quantile(0.75)
    IQR = Q3 - Q1
    low_limit = Q1 - (IQR * 1.5)
    high_limit = Q3 + (IQR * 1.5)

    filtered_entries = ((merged_df[col] >= low_limit) & (merged_df[col] <= high_limit)) & filtered_entries

merged_df = merged_df[filtered_entries]

print(f'Jumlah baris setelah memfilter outlier: {len(merged_df)}')
```

Outliers adalah nilai yang sangat berjauhan dengan data umumnya. Ini dapat mempengaruhi kinerja suatu algoritma atau model. Terdapat jumlah outliers pada dataset Credit_Card_Balance sebesar **3227965**. Hal ini perlu dilakukan handling untuk mendapatkan data yang rapi agar nantinya model dapat menghasilkan kinerja yang sangat baik. Jika tidak dilakukan penanganan ini maka dapat meningkatkan pengaruh negatif pada algoritma yang sensitif terhadp outliers. Setelah dilakukan handling outliers, maka didapatkan Jumlah baris setelah memfilter outlier sebesar **2213965**

### FEATURE ENGINEERING

Feature Engineering diperlukan untuk menambah kompleksitas data agar pada saat pemodelan didapatkan hasil dengan kinerja yang paling baik. Hal-hal yang dilakukan adalah melakukan penambahan feature baru dari feature-feature yang sudah ada. Berikut feature-feature yang perlu ditambahkan:

**Utilization Rate**          : Rasio antara jumlah kredit yang digunakan terhadap batas kredit. Contoh: AMT_BALANCE / AMT_CREDIT_LIMIT_ACTUAL.

**Regularization**          : Rasio antara jumlah yang harus dibayar terhadap umlah cicilan atau angsuran minimal yang dibayar secara berkala dalam jangka waktu tertentu. Contoh: AMT_PAYMENT_CURRENT / AMT_INST_MIN_REGULARITY.

## Handling Outlier

```
def outlierHandling(data, columns):
    for i in columns:
        Q1 = data[i].quantile(0.25)
        Q3 = data[i].quantile(0.75)
        IQR = Q3-Q1
        high_limit = Q3 + (1.5*IQR)
        low_limit = Q1 - (1.5*IQR)
    
        data = data[((data[i] >= low_limit) & (data[i] <= high_limit))]

    return data
```

```
features = ['MONTHS_BALANCE', 'CNT_INSTALMENT', 'CNT_INSTALMENT_FUTURE', 'SK_DPD', 'SK_DPD_DEF', 
            'remaining_instalment_pct', 'credit_duration', 'months_since_last_payment']
```

## More Outliers

```
# Define features with extreme skewness for outlier handling
outlier_features = ['SK_DPD', 'SK_DPD_DEF', 'credit_duration', 'months_since_last_payment']

# Cap outliers at the 1st and 99th percentiles
for feature in outlier_features:
    lower_cap = data[feature].quantile(0.01)
    upper_cap = data[feature].quantile(0.99)
    data[feature] = data[feature].clip(lower=lower_cap, upper=upper_cap)

# Verify that outliers are handled
print("Summary after outlier handling:")
print(data[outlier_features].describe())
```

```
# Existing features for log transformation
log_transform_features = ['CNT_INSTALMENT', 'CNT_INSTALMENT_FUTURE']

# Additional features based on skewness
additional_log_transform_features = ['SK_DPD', 'SK_DPD_DEF', 'credit_duration', 'months_since_last_payment']
log_transform_features.extend(additional_log_transform_features)

# Apply log transformation
for feature in log_transform_features:
    data[f"{feature}_LOG"] = np.log1p(data[feature])  # log1p to handle zero and negative values

# Summary to confirm transformations
print("Summary after log transformation:")
print(data[[f"{feature}_LOG" for feature in log_transform_features]].describe())
```
